```python
import matplotlib.pyplot as plt
```

```python
Platform_counts=df['Platform'].value_counts()
Platform_counts.plot(kind='bar',color='coral')
plt.title('Distribitions of platforms')
plt.xlabel('platform')
plt.ylabel('Count')
plt.show()
```

```python
sentiment_counts=df['Sentiment'].value_counts().head(10)
sentiment_counts.plot(kind='bar',color='coral')
plt.title('Top 10 Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
hours_counts= df['Hour'].value_counts().sort_index()
hours_counts.plot(kind='bar',color='coral')
plt.title(' Distribution of Hours')
plt.xlabel('Hour')
plt.ylabel('Count')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
years_counts= df['Year'].value_counts().sort_index()
years_counts.plot(kind='bar',color='coral')
plt.title(' Distribution of years')
plt.xlabel('year')
plt.ylabel('Count')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
df['Likes'].plot(kind='box',color='coral')
plt.title('box polt of likes')
plt.ylabel('values')
plt.show()
```

```python
df['Retweets'].plot(kind='hist',bins=20,color='coral',edgecolor='black')
plt.title('Histogram of Retweets')
plt.xlabel('Retweets')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Country']=df['Country'].astype(str).str.strip()
```

```python
country_counts=df['Country'].value_counts().head(10)
country_counts.plot(kind='bar',color='coral')
plt.title('Top 10 country Distribution')
plt.xlabel('country')
plt.ylabel('Count')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Retweets'],df['Likes'],color='coral',alpha=0.5,edgecolor='black')
plt.title('scatter plot:Retweets vs Likes')
plt.xlabel('Retweets')
plt.ylabel('Likes')
plt.show()
```

```python
month_counts=df['Month'].value_counts().sort_index()
plt.Figure(figsize=(8,6))
month_counts.plot(kind='bar',color='coral',edgecolor='black')
plt.title('distribution of months')
plt.xlabel('month')
plt.ylabel('count')
plt.xticks(rotation=0)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,6))
numeric_df=df.select_dtypes(include=['float64','int64'])
sns.heatmap(numeric_df.corr(),annot=True,cmap='coolwarm',fmt='.2f',linewidths=0.5)
plt.title('correlation heatmap')
plt.show()
```
